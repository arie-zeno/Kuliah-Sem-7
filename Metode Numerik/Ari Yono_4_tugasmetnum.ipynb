```python
import math
import numpy as np
import pandas as pd
from prettytable import PrettyTable
```

```python
def fungsi(x): #fungsi e^-x-2x^2+3x+3
    return round(math.exp(-x)-2*(x**2)+(3*x)+3, 5)
```

```python
def df(x): #function turunan
  return -math.exp(-x) - 4*x + 3
```

```python
def cari_selang_perubahan_tanda(fungsi, a, b, langkah): #function menentukan selang
  selang = []
  x_sebelumnya = a
  y_sebelumnya = fungsi(a)

  data = {'x': [], 'f(x)': [], 'Selang': []}
  label_selang = 1

  for x in np.arange(a+langkah, b+langkah, langkah):
    y = fungsi(x)
    data['x'].append(x)
    data['f(x)'].append(y)

    # Periksa perubahan tanda dan tambahkan label selang
    if y * y_sebelumnya < 0:
      selang.append((x_sebelumnya, x))
      data['Selang'][-1] = f"Selang {label_selang}"
      data['Selang'].append(f"Selang {label_selang}")
      label_selang += 1
    else:
      data['Selang'].append('-')

    x_sebelumnya = x
    y_sebelumnya = y

  df = pd.DataFrame(data)
  return df, selang
```

```python
def bisection(f, a, b, tol): #function metode bagi 2
  """
  Mencari akar suatu fungsi menggunakan metode bagi dua dan menampilkan tabel iterasi.

  Args:
    f: Fungsi yang akan dicari akarnya.
    a: Batas bawah interval.
    b: Batas atas interval.
    tol: Toleransi (error maksimum yang diizinkan).

  Returns:
    Aproksimasi akar fungsi.
  """

  if f(a) * f(b) >= 0:
    print("Fungsi tidak memiliki akar pada interval yang diberikan.")
    return None

  table = PrettyTable(["Iterasi", "a", "b", "f(a)", "f(b)", "t", "f(t)", "Toleransi"])

  iterasi = 1
  true_count = 0
  last_true_t = None

  while True:
    t = (a + b) / 2
    ft = f(t)
    bandingan_toleransi = abs(ft) <= tol
    table.add_row([iterasi, f"{a:.4f}", f"{b:.4f}", f"{f(a):.4f}", f"{f(b):.4f}", f"{t:.4f}", f"{ft:.4f}", bandingan_toleransi])

    if bandingan_toleransi:
      true_count += 1
      last_true_t = t
      if true_count >= 3:
        print(table)
        print(f"\nAkar ditemukan setelah {iterasi} iterasi: {last_true_t:.4f}")
        return last_true_t
    else:
      true_count = 0
      last_true_t = None

    if f(a) * ft < 0:
      b = t
    else:
      a = t

    iterasi += 1
```

```python
def posisi_palsu(f, a, b, tol): #function metode palsu
  """
  Mencari akar suatu fungsi menggunakan metode posisi palsu.

  Args:
    f: Fungsi yang akan dicari akarnya.
    a: Batas bawah interval.
    b: Batas atas interval.
    tol: Toleransi (error maksimum yang diizinkan).

  Returns:
    Aproksimasi akar fungsi.
  """

  if f(a) * f(b) >= 0:
    print("Fungsi tidak memiliki akar pada interval yang diberikan.")
    return None

  table = PrettyTable(["Iterasi", "a", "b", "f(a)", "f(b)", "c", "f(c)", "Toleransi"])

  iterasi = 1
  while (b-a)/2 > tol:
    # Perbedaan utama: Hitung c menggunakan metode posisi palsu
    c = b - (f(b)*(b-a)) / (f(b)-f(a))
    fc = f(c)

    # Hitung toleransi
    toleransi = abs(fc) <= tol

    table.add_row([iterasi, f"{a:.4f}", f"{b:.4f}", f"{f(a):.4f}", f"{f(b):.4f}", f"{c:.4f}", f"{fc:.4f}", toleransi])

    if fc == 0:
      break
    elif f(c)*f(a) < 0:
      b = c
    else:
      a = c

    iterasi += 1

  print(table)
  print(f"\nAkar ditemukan setelah {iterasi} iterasi: {c:.4f}")
  return c
```

```python
def newton_raphson(f, df, x0, tol): #function metode newton-raphson
    """
    Mencari akar suatu fungsi menggunakan metode Newton-Raphson.

    Args:
        f: Fungsi yang akan dicari akarnya.
        df: Turunan pertama dari fungsi f.
        x0: Tebakan awal akar.
        tol: Toleransi (error maksimum yang diizinkan).

    Returns:
        Aproksimasi akar fungsi.
    """

    table = PrettyTable(["Iterasi", "an", "f(an)", "f'(an)", "xn+1", "Toleransi"])

    n = -1
    an = x0
    toleransi_count = 0  # Menghitung jumlah toleransi True berturut-turut
    while toleransi_count < 3:
        fn = f(an)
        dfn = df(an)
        xn1 = an - fn / dfn

        # Hitung toleransi
        toleransi = abs(xn1-an) <= tol
        toleransi_count = toleransi_count + 1 if toleransi else 0  # Reset jika False

        table.add_row([n+1, f"{an:.4f}", f"{fn:.4f}", f"{dfn:.4f}", f"{xn1:.4f}", toleransi])

        an = xn1
        n += 1

    print(table)
    print(f"\nAkar ditemukan setelah {n} iterasi: {xn1:.4f}")
    return xn1
```

```python
def secant_method(f, a, b, tol): #function metode secant
    """
    Mencari akar suatu fungsi menggunakan metode secant.

    Args:
        f: Fungsi yang akan dicari akarnya.
        a: Tebakan awal pertama.
        b: Tebakan awal kedua.
        tol: Toleransi (error maksimum yang diizinkan).

    Returns:
        Aproksimasi akar fungsi.
    """

    table = PrettyTable(["Iterasi", "xn-1", "xn", "f(xn-1)", "f(xn)", "xn+1", "Toleransi"])
    n = 0
    xn = b
    xn_1 = a
    while abs(xn - xn_1) > tol:
        xn_2 = xn - f(xn) * (xn - xn_1) / (f(xn) - f(xn_1))
        table.add_row([n+1, f"{xn_1:.4f}", f"{xn:.4f}", f"{f(xn_1):.4f}", f"{f(xn):.4f}", f"{xn_2:.4f}", abs(xn_2 - xn) < tol])
        xn_1 = xn
        xn = xn_2
        n += 1
    
    print(table)
    print(f"\nAkar ditemukan setelah {n} iterasi: {xn_2:.4f}")
    return xn_2
```

```python
# mencari selang dengan batas awal -5 sampai 5, toleransi 0,0001
a = -5
b = 5
tol = 0.0001
langkah = 1
hasil, selang_list = cari_selang_perubahan_tanda(fungsi, a, b, langkah)
print(hasil)

# Menampilkan hasil selang
for i, selang in enumerate(selang_list, start=1):
  print(f"Selang {i}: ({selang[0]:.2f}, {selang[1]:.2f})")
```
